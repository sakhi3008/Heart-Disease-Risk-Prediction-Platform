### Step 1: Load Gold Data

```python
gold_df = spark.read.table("workspace.default.gold_heart_disease")
display(gold_df)
gold_df.printSchema()
```

### Step 2: Basic Dataset Overview

```python
gold_df.count()
```

### Step 3: Descriptive Statistics

```python
gold_df.select(
    "age",
    "resting_blood_pressure",
    "serum_cholesterol",
    "max_heart_rate"
).describe().show()
```

### Step 4: Heart Disease Distribution

```python
gold_df.groupBy("risk_label").count().show()
```

### Step 5: Gender-wise Risk Analysis

```python
gold_df.groupBy("gender", "risk_label").count().show()
```

### Step 6: Age Group vs Risk

```python
from pyspark.sql.functions import when

age_group_df = gold_df.withColumn(
    "age_group",
    when(gold_df.age < 40, "Young")
    .when(gold_df.age < 60, "Middle Age")
    .otherwise("Senior")
)

age_group_df.groupBy("age_group", "risk_label").count().show()
```

### Step 7: Correlation Analysis

```python
age_group_df.stat.corr("serum_cholesterol", "heart_disease_flag")
```
