### SQL Analytics & Risk Dashboards

In this notebook, we analyze heart disease risk predictions generated by the ML model.
Using SQL, we create insights and aggregations that can be visualized in dashboards
to support clinical and business decision-making.


### Step 1: Load Final Predictions Table

```python
%sql
SELECT * 
FROM workspace.default.gold_heart_risk_predictions
LIMIT 10;
```

### Step 2: Risk Distribution

```python
%sql
SELECT
  risk_level,
  COUNT(*) AS patient_count
FROM workspace.default.gold_heart_risk_predictions
GROUP BY risk_level
ORDER BY patient_count DESC;
```

### Step 3: Average Risk Probability by Risk Level

```python
%sql
SELECT
  risk_level,
  ROUND(AVG(risk_probability), 3) AS avg_risk_probability
FROM workspace.default.gold_heart_risk_predictions
GROUP BY risk_level;
```

### Step 4: Join with Patient Attributes

```python
%sql
SELECT
  g.age,
  g.gender_label,
  g.serum_cholesterol,
  g.resting_blood_pressure,
  p.risk_level,
  p.risk_probability
FROM workspace.default.gold_heart_risk_predictions p
JOIN workspace.default.gold_heart_disease g
ON p.patient_id = g.patient_id;
```

### Step 5: Risk by Gender

```python
%sql
SELECT
  gender_label,
  risk_level,
  COUNT(*) AS patient_count
FROM workspace.default.gold_heart_risk_predictions p
JOIN workspace.default.gold_heart_disease g
ON p.patient_id = g.patient_id
GROUP BY gender_label, risk_level
ORDER BY gender_label, patient_count DESC;
```

### Step 6: Risk by Age Group

```python
%sql
SELECT
  g.age_group,
  p.risk_level,
  COUNT(*) AS patient_count
FROM workspace.default.gold_heart_risk_predictions p
JOIN workspace.default.gold_heart_feature_details g
ON p.patient_id = g.patient_id
GROUP BY g.age_group, p.risk_level
ORDER BY g.age_group;
```

### Step 7: High-Risk Patient Identification

```python
%sql
SELECT
  g.age,
  g.gender_label,
  g.serum_cholesterol,
  g.resting_blood_pressure,
  p.risk_probability
FROM workspace.default.gold_heart_risk_predictions p
JOIN workspace.default.gold_heart_disease g
ON p.patient_id = g.patient_id
WHERE p.risk_level = 'High Risk'
ORDER BY p.risk_probability DESC;
 
```
