## Bronze Layer – Raw Clinical Data

### Step 1: Load Raw Heart Disease Data

In this step, we read the Cleveland Heart Disease dataset from the Databricks table into a Spark DataFrame for further processing.


```python
df_raw = spark.read.table("workspace.default.heart")
df_raw.show(5)
df_raw.printSchema()
```

### Step 2: Add Ingestion Timestamp

We add an ingestion timestamp column to track when the data entered the Lakehouse.


```python
from pyspark.sql.functions import current_timestamp

bronze_df = df_raw.withColumn("ingestion_timestamp", current_timestamp())
bronze_df.show(5)
```

### Step 3: Store Data in Bronze Layer as Delta Table

The cleaned raw data is stored in Delta format to enable ACID transactions, versioning, and scalable analytics.

```python
bronze_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.bronze_heart_disease")
```

### Step 4: Verify Bronze Table

Confirm that the raw heart disease data is successfully stored as a Delta table in the Bronze layer.

```python
%sql
SELECT * 
FROM workspace.default.bronze_heart_disease 
LIMIT 10;
```
